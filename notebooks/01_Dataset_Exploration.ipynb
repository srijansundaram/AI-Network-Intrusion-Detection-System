# Stage 1 - Dataset Inventory

## Objective

Before preprocessing or model development, it is important to understand the composition of the CICIDS-2017 dataset.

In this stage, we identify all available CSV files, inspect their sizes, determine the number of rows and columns, and understand what each file represents.

This provides an overview of the dataset and helps verify that all required files are available.

```python
csv_files = sorted(
    [file for file in os.listdir(DATA_PATH) if file.endswith(".csv")]
)

print(f"Total CSV files: {len(csv_files)}\n")

for file in csv_files:
    print(file)
```

```python
inventory = []

for file in csv_files:

    path = os.path.join(DATA_PATH, file)

    df = pd.read_csv(path, low_memory=False)

    inventory.append({
        "File": file,
        "Rows": df.shape[0],
        "Columns": df.shape[1],
        "Size (MB)": round(os.path.getsize(path)/(1024*1024),2)
    })

inventory_df = pd.DataFrame(inventory)

inventory_df
```

## Observation

The CICIDS-2017 dataset consists of multiple CSV files collected across different weekdays.

Each CSV file represents network traffic captured during a particular time period and contains flow-based features extracted using CICFlowMeter.

The table above summarizes the number of samples, feature dimensions, and storage size of each dataset.

```python
total_rows = inventory_df["Rows"].sum()

print(f"Total Samples : {total_rows:,}")
```

```python
total_size = inventory_df["Size (MB)"].sum()

print(f"Dataset Size : {total_size:.2f} MB")
```

## Key Observations

1) The CICIDS-2017 dataset consists of 8 CSV files collected over five working days (Monday to Friday), representing different network traffic scenarios.
2) Each CSV file contains either normal network traffic, a specific type of cyber attack, or a combination of both, depending on the day and session in which the data was collected.
3) The dataset contains millions of network flows, where each row represents a network flow rather than an individual packet. Therefore, the total number of rows corresponds to the total number of communication flows captured during that session.
4) Every CSV file contains 79 features, indicating a consistent dataset structure. This uniform schema makes it easier to merge all files into a single dataset during preprocessing.
5) The number of samples and storage size varies across different files, indicating that network traffic volume changes depending on the day and the attack scenarios being simulated.
6) The dataset has been systematically organized by separating attack scenarios into different days and time periods (Morning/Afternoon), making it easier to analyze individual attacks as well as evaluate machine learning models under different network conditions.

# Stage 3 - Feature Exploration

## Objective

Explore the different categories of features present in the CICIDS-2017 dataset and understand how they describe network communication for intrusion detection.

```python
feature_list = list(df.columns)

for i, feature in enumerate(feature_list, start=1):
    print(f"{i}. {feature}")
```

```python
print("Total Features :", len(df.columns)-1)
print("Target Column :", df.columns[-1])
```

The CICIDS-2017 dataset contains 78 features that can be grouped into logical categories based on the type of network information they represent. The following table summarizes these feature groups and their significance.

| Feature Category               | Purpose                                                                 | Example Features                                           |
| ------------------------------ | ----------------------------------------------------------------------- | ---------------------------------------------------------- |
| **Network Identification**     | Identifies the destination service or endpoint                          | Destination Port                                           |
| **Flow Statistics**            | Describes the overall network communication flow                        | Flow Duration, Flow Bytes/s, Flow Packets/s                |
| **Packet Statistics**          | Measures packet counts and packet sizes                                 | Total Fwd Packets, Packet Length Mean, Average Packet Size |
| **Forward & Backward Traffic** | Separately analyzes client-to-server and server-to-client communication | Fwd Packet Length Mean, Bwd IAT Mean                       |
| **TCP Flag Features**          | Represents TCP connection behaviour and control flags                   | SYN Flag Count, ACK Flag Count, FIN Flag Count             |
| **Bulk & Window Features**     | Describes bulk data transfer and TCP window characteristics             | Init_Win_bytes_forward, Fwd Avg Bytes/Bulk                 |
| **Activity Features**          | Measures active periods of network communication                        | Active Mean, Active Max                                    |
| **Idle Features**              | Measures inactive periods between communications                        | Idle Mean, Idle Max                                        |


## Key Observations

1. The dataset contains 78 input features for each network communication flow along with one target label that indicates whether the traffic is BENIGN or belongs to a specific attack category.

2. The 78 features can be grouped into eight logical categories: Network Identification, Flow Statistics, Packet Statistics, Forward & Backward Traffic, TCP Flag Features, Bulk & Window Features, Activity Features, and Idle Features.

3. Network Identification features help identify the destination service being accessed through the destination port number.

4. Flow Statistics describe the overall behaviour of a communication flow, such as its duration, data transfer rate, and packet transmission rate.

5. Packet Statistics provide information about the number of packets exchanged and their size characteristics during a network flow.

6. Forward and Backward Traffic features separately analyze communication from the client to the server and from the server back to the client.

7. Bulk & Window features describe bulk data transfer behaviour and TCP window characteristics, which can help identify abnormal traffic patterns.

8. Activity features measure the duration and statistical characteristics of active communication periods within a network flow.

9. Idle features measure the periods when no communication occurs, helping describe the inactive behaviour of a network flow.

10. Together, these features capture different aspects of network communication, enabling machine learning algorithms to distinguish between normal traffic and various cyber attacks without inspecting the actual packet payload.

# Stage 5 - Missing Value Analysis

## Objective

Check whether the dataset contains missing values.


```python
missing = df.isnull().sum()

missing[missing > 0]
```

```python
df.isnull().sum().sum()
```

```python
df[df["Flow Bytes/s"].isnull()]
```

```python
df[df["Flow Bytes/s"].isnull()][[" Flow Duration", "Flow Bytes/s"]]
```

## Key Observations

1. The Monday dataset contains **64 missing values**, all of which are present in the `Flow Bytes/s` feature.

2. Further investigation revealed that these missing values are not caused by incomplete data collection but are the result of a mathematical calculation.

3. Some network flows have a **Flow Duration** of zero. Since `Flow Bytes/s` is calculated by dividing the total number of bytes by the flow duration, division by zero results in undefined (`NaN`) values.

4. The corresponding `Flow Packets/s` values for these records are `Infinity`, confirming that the issue is caused by division by zero.

5. These anomalous values will be handled during the preprocessing stage to ensure that the machine learning models are trained on clean and consistent data.
